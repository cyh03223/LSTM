```python
import subprocess
import os
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import cont2discrete, lti, dlti, dstep
import scipy.integrate
import random

# Set up environment variables for different GPUs
def set_env(gpu_id):
    env = os.environ.copy()
    env["CUDA_VISIBLE_DEVICES"] = str(gpu_id)
    return env

c_r = 100 #fill in the c_r value ### same thing

# Define models and logs ***
models = {
    "JLSTM": {"model number": "2", "gpu": 1, "log": f"jlstm_ks_result_cr_{str(c_r).replace('.', '_')}.txt"},
    "L-JLSTM": {"model number": "1", "gpu": 2, "log": f"l_jlstm_ks_result_cr_{str(c_r).replace('.', '_')}.txt"},
    "EKF": {"model number": "3", "gpu": 3, "log": f"ekf_ks_result_cr_{str(c_r).replace('.', '_')}.txt"}
}

# Start models
processes = {}
start_times = {}

for model, config in models.items():
    start_times[model] = time.time()
    processes[model] = subprocess.Popen(
        ["python", "ks_noise_comparison.py", "--model_id", config["model number"]],
        stdout=open(config["log"], "w"),
        stderr=subprocess.STDOUT,
        env=set_env(config["gpu"])
    )

print("Models are running in parallel. Check logs for details.")
```
